```python
import pandas as pd
import boto3
import os

from io import BytesIO
from pathlib import Path
from dotenv import load_dotenv
```

```python
path = Path().cwd().parent.parent / ".env"
load_dotenv(dotenv_path=path)
```

```python
s3 = boto3.client(
    "s3",
    endpoint_url="http://localhost:9010",
    aws_access_key_id=os.environ["S3_ACCESS_KEY"],
    aws_secret_access_key=os.environ["S3_SECRET_KEY"],
)
```

```python
data = s3.get_object(
    Bucket="datalake-silver",
    Key="arxiv/arxiv_file_20260920_00-11-11.parquet")["Body"].read()
```

```python
df = pd.read_parquet(BytesIO(data))
```

```python
df.head()
```

```python




df = pd.read_xml(
    BytesIO(data),
    xpath="//atom:entry",
    namespaces={"atom": "http://www.w3.org/2005/Atom"},
)

print(df.head())
print(df.columns)

s3.put_object(
    Bucket="datalake-silver",
    Key="arxiv/arxiv_file_20260920_00-11-11.parquet",
    Body=df.to_parquet(index=False)
)
```
